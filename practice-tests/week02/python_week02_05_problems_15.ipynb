```python
#12
a, b = map(int,input().split())
quotient = 0
rest = float("-inf")
gcd = 0
while rest != 0 :
    if a > b :
        quotient = a // b 
        rest = a % b
        a = b
        b = rest
        gcd = a
    elif a < b :
        quotient = b // a 
        rest = b % a
        b = a
        a = rest
        gcd = b
    else :
        gcd = a
        break
    
print(f"최대공약수: {gcd}")
```

```python
#13
N13 = int(input())

trial = 0
while N13 != 1 :
    if N13 == 0 :
        trial = float("inf")
        break
    elif N13 % 2 == 0 :
        N13 = N13 // 2
    else :
        N13 = 3*N13 + 1
    trial += 1

print(f"횟수: {trial}")
```

```python
#14
N14 = int(input())
cumulative_sum = 0
current_n = 0
if N14 <= 0 :
    print("1 이상의 정수 입력 필요")
else:
    while cumulative_sum < N14 :
        current_n += 1
        cumulative_sum += current_n


print(f"마지막 수: {current_n}")
print(f"누적합: {cumulative_sum}")

```
