```python
#14
m_score, attend_ratio, warn = map(float, input().split())

warn = int(warn)

if m_score >= 85 and attend_ratio >= 90 and warn == 0 :
    print("장학금 대상입니다.")
else:
    print("장학금 대상이 아닙니다.")
```

```python
#15
acc_ex, time_infer, exis_error = input().split()
acc_ex, time_infer = map(float, (acc_ex,time_infer))
exis_error = exis_error.upper()

if exis_error == "Y" :
    state = "실험 실패"
elif exis_error == "N" and acc_ex >= 95 and time_infer <= 1.0:
    state = "배포 후보" 
elif exis_error == "N" and acc_ex >= 90:
    state = "추가 최적화 필요"
else :
    state = "성능 부족"
    
print(f"최종 상태: {state}")
    
    
```
