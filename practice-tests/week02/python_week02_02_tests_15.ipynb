```python
#1
# 첫번째 방법
a, b, c = map(int, input().split())

if a > b and a > c :
    if b > c :
        mid_num = b
    else :
        mid_num = c
elif b > a and b > c :
    if a > c :
        mid_num = a
    else :
        mid_num = c
elif c > a and c > b :
    if a > b :
        mid_num = a
    else :
        mid_num = b
        
print("중앙값:",mid_num)
# 두번째 방법
a, b, c = map(int, input().split())

max_num = max(a,b,c)

if max_num == a and b > c :
    mid_num = b if b > c else  c
elif max_num == b and a > c :
    mid_num = a if a > c else  c
elif max_num == c and a > b :
    mid_num = a if a > b else  b
    
print("중앙값:",mid_num)   
    
     
```

```python
#4
height, weight = map(float, input().split())

height = height/100

bmi = weight / (height**2)
print(f"BMI: {bmi:.2f}")

if bmi >= 25 :
    print("판정: 비만")
elif bmi >= 23 :
    print("판정: 과체중")
elif bmi >= 18.5 :
    print("판정: 정상")
else :
    print("판정: 저체중")

```

```python
#14
month, day = map(int, input().split())

is_day_right = ((month in [1,3,5,7,8,10,12]) and 1 <= day <= 31) or ((month in [2,4,6,9,11]) and 1 <= day <= 30)

if month in (3,4,5) and is_day_right : 
    print("봄")
elif month in (6,7,8) and is_day_right :
    print("여름")
elif month in (9,10,11) and is_day_right :
    print("가을")
elif month in (12,1,2) and is_day_right:
    print("겨울")
else:
    print("날짜를 제대로 입력하시요")
```
