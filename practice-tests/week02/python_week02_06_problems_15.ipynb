```python
#3
trial_3 = 0
while trial_3 < 3 :
    pw = input()
    if pw == "python123" :
        print("로그인 성공")
        break
    else :
        trial_3 += 1
        print(f"비밀번호 오류({trial_3}/3)")
if trial_3 == 3 :
    print("계정 잠금")
```

```python
#4
num_4 = 0
available_num = 0
sum_score = 0
while num_4 < 5 :
    score =  float(input())
    num_4 += 1 
    if 0 <= score <= 100 :
        available_num += 1
    else :
        continue
    sum_score += score  
print(f"유효 개수: {available_num}")
print(f"합계: {sum_score}")
    
```

```python
#6
while True :
    N6 = int(input())
    if N6 >= 2 :
        break
    else:
        print("2 이상의 정수를 입력하시오")

i = 2
while i < N6 :
    if N6 % i == 0 :
        print(f"첫 번째 약수: {i}")
        break
    i += 1
if i == N6 :
    print("소수")
```

```python
#10
couple_num = 0
sum_10 = 0
while couple_num < 5 :
    try:
        N10 = int(input())
        if N10 % 2 == 0 :
            couple_num += 1
            sum_10 += N10
        
    except ValueError:
        print("정수만 입력하세요.")
print(f"짝수 합: {sum_10}")
        
```

```python
#11
value_num = 0
sum_value = 0
while True :
    try :
        sensor_value = float(input())
        if sensor_value == -999 :
            break
        elif 0 <= sensor_value <= 100 :
            value_num += 1
            sum_value += sensor_value
        
    except ValueError:
        print("Wrong Value")
if value_num > 0:
    print(f"평균: {sum_value/value_num:.2f}")
else:
    print("유효한 데이터가 없습니다.")
```

```python
#12
while True:
    N12 = int(input())
    if N12 <= 0 :
        print("자연수를 입력하세요")
        continue
    break

i = 2
prime_num = 0
while prime_num < N12 :
    j = 2
    is_prime_num = True
    while j < i :
        if i % j == 0 :
            is_prime_num = False
            break
        j += 1
    if is_prime_num == True :
        prime_num += 1
        print(i, end= " ")
    i += 1
    
    
    
```

```python
#13
correct_num = 37
trial_13 = 0
while True :
    try:
        N13 = int(input())
        trial_13 += 1
        if N13 > correct_num :
            print("낮춰주세요")
        elif N13 < correct_num:
            print("높여주세요")
        else:
            print(f"정답! {trial_13}회")
            break
    except ValueError:
        print("정수를 입력하세요")
```

```python
#14
list_14 = list(map(int,input().split()))

i = 0

while i < len(list_14) :
    if list_14[i] == 1 :
        print(f"첫 이상 프레임: {i}")
        break
    i += 1
if i == len(list_14) :
    print("-1")
```

```python
#15
sum_even = 0
num_odd = 0
while True:
    try:
        N15 = int(input())
        
        if N15 == 0 :
            break
        
        if N15 < 0 :
            continue
        elif N15 % 2 == 0 :
            sum_even += N15
        elif N15 % 2 == 1 :
            num_odd += 1
        
    except ValueError:
        print("정수를 입력하세요")
        
print(f"짝수 합: {sum_even}")
print(f"홀수 개수: {num_odd}")
```
