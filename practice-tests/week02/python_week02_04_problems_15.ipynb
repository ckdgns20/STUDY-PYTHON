```python
#4
N4 = int(input())

fac = 1

if N4 >= 0 :
    for i in range(1,N4+1):
        fac *= i
    print(f"{N4}! = {fac}")
else :
    print("0이상의 정수를 입력하시요")
    
    
```

```python
#5
list5 = list(map(int,input().split()))

min_num = list5[0]
min_index = 0
for i in range(len(list5)):
    if min_num > list5[i] :
        min_num = list5[i]
        min_index = i 
        
print(f"최솟값: {min_num} \n위치: {min_index}")
```

```python
#6
list6 = list(map(int,input().split()))

max_num = list6[0]
second_num = float('-inf')

for i in range(len(list6)):
    if max_num < list6[i] :
        second_num = max_num
        max_num = list6[i]
    elif second_num < list6[i] < max_num:
        second_num = list6[i]

print(f"두 번째 큰 수: {second_num}")
        
```

```python
#11
list11 = []
for _ in range(2):
    list11.append(list(map(int,input().split())))

sum_total = 0
# 리스트는 for문에 들어가면 큰묶음 먼저 인식(가로줄 먼저)
for row in list11:
    for item in row:
        sum_total += item
        
print(f"전체 합: {sum_total}")
```

```python
#12
list12 = []
for _ in range(2):
    list12.append(list(map(int,input().split())))
    
sum_by_row = 0

for i,row in enumerate(list12) :
    for item in row :
        sum_by_row += item
    print(f"{i+1}행 합: {sum_by_row}")
    sum_by_row = 0
```

```python
#13
A,B = map(int, input().split())
# break는 제일 안쪽 반복문을 벗어남(조건문과 상관없음)
for i in range(A,B+1):
    isprime = True
    for j in range(2,i) :
        if i % j == 0 :
            isprime = False
            break
    if isprime == True :
        print(i, end=" ")     
```

```python
#14
list14 = list(map(int,input().split()))

interval = 1
for i in range(0,len(list14)-1):
    start_num = list14[i]
    end_num = list14[i+1]
    if start_num < end_num :
        interval += 1
    else :
        break
    
print(f"초기 증가 길이: {interval}")
```

```python
#15
N,W,S = map(int,input().split())

for i in range(0,N,S):
    full_idx = i + W 
    if full_idx <= N :
        print(i,end=' ')
    else :
        break
```
