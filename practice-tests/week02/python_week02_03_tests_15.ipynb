```python
#2
N2 = int(input())

for i in range(N2, 0, -1):
    print(i, end= ' ')
```

```python
#3
N3_list = list(map(int,input().split()))

startn = N3_list[0]
endn = N3_list[1]

for i in range(startn, endn+1):
    if i % 2 == 0 :
        continue
    print(i, end=' ')
```

```python
#4
N4 = int(input())

sum_total = 0

for i in range(1,N4+1):
    sum_total += i
    
print(f"합계: {sum_total}")
```

```python
#5
N5 = int(input())

sum_couple = 0

for i in range(1,N5+1):
    if i % 2 != 0 :
        continue
    sum_couple += i
    
print(f"짝수 합: {sum_couple}")
```

```python
#6
N6, times = map(int,(input().split()))

num_times = 0

for i in range(1,N6+1):
    if i % times != 0:
        continue
    num_times += 1
    
print(f"개수: {num_times}")
```

```python
#9
L9 = list(map(int,input().split()))

sum_total9 = 0

for i in L9:
    sum_total9 += i
    
print(f"합계: {sum_total9} \n평균: {sum_total9/len(L9):.2f}")
```

```python
#10
L10 = list(map(int,input().split()))

num_plus = 0

for i in L10:
    if i > 0 :
        num_plus += 1
    else:
        continue
    
print(f"양수 개수: {num_plus}")
```

```python
#11
L11 = list(map(int, input().split()))

max_num = L11[0]

for i in L11:
    if i > max_num :
        max_num = i
    else :
        continue
    
print(f"최댓값: {max_num}")
```

```python
#12
L12 = list(map(int,input().split()))

for i in range(len(L12)):
    print(f"{i}: {L12[i]}")
```
