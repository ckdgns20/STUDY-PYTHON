```python
#1
def show_name(name):
    print(f"이름: {name}")
    
show_name(input())
```

```python
#2
def show_numbers(a,b):
    print(a,b)

a,b = map(int,(input().split()))
show_numbers(a,b)
```

```python
#3
def print_sum(a, b):
    print(f"합계: {a+b}")
    
a,b = map(int,input().split())
print_sum(a,b)
```

```python
#4
def show_rectangle(width, height):
    dimensions = width*height
    
    print(f"가로: {width}")
    print(f"세로: {height}")
    print(f"넓이: {dimensions}")
    
a,b = map(int,input().split())
show_rectangle(a,b)
```

```python
#5
def greet(name, massage='안녕하세요'):
    print(f"{massage}, {name}님")
    
greet('창훈')
greet('창훈','반갑습니다')
```

```python
#6
def show_length(value, unit='mm'):
    print(value, unit)
    
show_length(25)
show_length(3.5,'cm')
```

```python
#7
def show_range(start, end):
    for i in range(start,end+1):
        print(i,end=" ")

start, end = map(int,(input().split()))
show_range(start,end)
```

```python
#8
def show_window(start, window=64):
    end = start + window - 1
    print(f"시작: {start}")
    print(f"끝: {end}")

show_window(100)
show_window(100,32)
```

```python
#9
def show_sampling(fps, stride=1):
    print(f"FPS: {fps}, stride: {stride}")
    
show_sampling(60)
show_sampling(60, 16)
```

```python
#10
def check_score(score, threshold=90):
    if score >= threshold :
        print("통과")
    else:
        print("미달")
        
check_score(88, 85)
```

```python
#11
def chek_sensor(value, low, high):
    if value >= low and value <= high:
        print('정상')
    else:
        print('이상')

value, low, high = map(float, input().split())

chek_sensor(value,low,high)
```

```python
#12
def show_resolution(width=1280, height= 720):
    print(f"{width} x {height}")

show_resolution()    
show_resolution(1920,1080)
```

```python
#13
def show_prediction(label, score, threshold=0.8):
    if score >= threshold :
        print(f"{label}: 확정")
    else:
        print(f"{label}: 보류")
        
show_prediction('normal',0.91)
```

```python
#14
def show_shipping(weight, base_fee=3000, heavy_limit=5):
    fee = base_fee
    if weight > heavy_limit :
        fee = base_fee + 2000
        print(f"배송비: {fee}")
    else:
        print(f"배송비: {fee}")
        
show_shipping(7)
```

```python
#15
def show_experiment(name, fps=60, window=64, stride=16):
    print(f"{name} | fps={fps} | window={window} | stride={stride}")
    
show_experiment('catheter')
```
