```python
#4
def calc_average(values):
    mean = sum(values)/len(values)
    return mean

input4 = list(map(int,input().split()))
output4 = calc_average(input4)
print(f"{output4:.1f}")
```

```python
#5
def get_min_max(values):
    return min(values), max(values)

input5 = list(map(int,input().split()))
min5 , max5 = get_min_max(input5)

print(f"최솟값: {min5}")
print(f"최댓값: {max5}")
```

```python
#6
def divide(a, b):
    return a // b , a % b
input6a, input6b = list(map(int,input().split()))
share, remain = divide(input6a,input6b)

print(f"몫: {share}")
print(f"나머지: {remain}")
```

```python
#7
def get_window(start, window=64):
    return start, start + window - 1

print(*get_window(100))

 
```

```python
#8
def check_score(score):
    wrong_score = "잘못된 점수" 
    correct_score = "정상 점수"
    if 0 > score or 100 < score :
        return wrong_score
    else :
        return correct_score
    
input8 = int(input())
print(check_score(input8))
```

```python
#9
def get_grade(score):
    if score >= 90 :
        return 'A'
    elif score >= 80 :
        return 'B'
    elif score >= 70 :
        return 'C'
    elif score >= 60 :
        return 'D'
    else :
        return 'F'
    
print(get_grade(float(input())))
```

```python
#14
def check(score, threshold):
    if score >= threshold:
        return True
    else:
        return False
    
print(check(0.91,0.95))
```

```python
#15
def summarize(scores):
    min_acc = min(scores)
    max_acc = max(scores)
    mean_acc = sum(scores)/len(scores)
    return mean_acc, min_acc, max_acc
input_acc = list(map(float, input().split()))

print("평균: {}, 최소: {}, 최대: {}".format(*summarize(input_acc)))
```
