```python
#1
def get_positive(values):
    result = []
    for i in values:
        if i > 0 :
            result.append(i)
    
    return result

input1 = list(map(int, input().split()))
result1 = get_positive(input1)

print(result1) 
```

```python
#2
def get_even(values):
    result = []
    for i in values:
        if i % 2 == 0 :
            result.append(i)
            
    return result

input2 = list(map(int, input().split(",")))
result2 = get_even(input2)

print(result2)
```

```python
#3
def summarize(values):
    max_value = max(values)
    min_value = min(values)
    mean_value = sum(values)/len(values)
    return mean_value, min_value, max_value

input3 = [10,20,30,40]
print("평균: {}, 최소: {}, 최대: {}".format(*summarize(input3)))
```

```python
#4
def filter_by_threshold(values, threshold):
    result4 = []
    for i in values:
        if i >= threshold :
            result4.append(i)
    return result4

input4_val, input4_thr = [0.72,0.91,0.83,0.95], 0.85
result4 = filter_by_threshold(input4_val, input4_thr)

print(result4)
```

```python
#5
def find_anomaly_frames(states):
    result5 = []
    for idx , i in enumerate(states):
        if i == 1 :
            result5.append(idx)
            
    return result5

input5 = [0,1,0,0,1,1]
result5 = find_anomaly_frames(input5)

print(result5)
```

```python
#6
def count_labels(labels):
    result6 = {}
    for label in labels:
        if label not in result6:
            result6[label] = 1
        else:
            result6[label] += 1
    return result6
        
input6 = ['normal','defect','normal','warning','normal']
result6 = count_labels(input6)       

print(result6)
```

```python
#7
def select_classes(scores, threshold):
    result7 = []
    for key, value in scores.items():
        if value > threshold :
            result7.append(key)
            
    return result7

input7_scores, input7_thr = {'A':0.91,'B':0.72,'C':0.88}, 0.85
result7 = select_classes(input7_scores, input7_thr)

print(result7)
```

```python
#8
def get_best_class(scores):
    
    best_class = next(iter(scores))
    best_score = scores[best_class]
    for key, value in scores.items():
        if value > best_score :
            best_class = key
            best_score = value
        
    return best_class, best_score

input8 = {'normal':0.82,'highspeed':0.94,'hightemp':0.89}
best_class, best_score = get_best_class(input8)

print(best_class, best_score)
```

```python
#10
def get_missing_ids(all_ids, done_ids):
    return all_ids- done_ids
input10a = {1,2,3,4,5}
input10b = {1,3,5}

print(get_missing_ids(input10a,input10b))
```

```python
#11
def unique_sorted(values):
    
    return sorted(set(values))

input11 = [4,2,4,1,2,3]

print(unique_sorted(input11))
    
```

```python
#12 --> 채점 시 틀렸다고 표시해줘. 구글링해서 작성한 코드야.
def extract_accuracy(results):
    return {model : metrics['acc'] for model, metrics in results.items()}
input12 = {'A':{'acc':0.95,'ms':0.5}, 'B':{'acc':0.97,'ms':0.7}}

print(extract_accuracy(input12))
```

```python
#13
def find_candidates(results, min_acc, max_ms):
    candidates_list = []
    
    for key, values in results.items():
        if (values['acc'] >= min_acc) and (values['ms'] <= max_ms):
            candidates_list.append(key)

    return candidates_list
    

input13 = input12
min_acc, max_ms = map(float, input().split())

print(find_candidates(input13,min_acc,max_ms))
```

```python
#14
def window_average(values, window=1):
    window_mean_list = []
    for i in range(0,len(values),window):
        window_mean_list.append(sum(values[i:i+window])/window)
    
    return window_mean_list

input14_val = [1,2,3,4,5,6]
window = 2

print(window_average(input14_val,window))
        
        
```

```python
#15
def make_summary(acc_list, time_list):
    summary_dict = {
        'avg_acc': sum(acc_list)/len(acc_list),
        'best_acc' : max(acc_list),
        'avg_time' : sum(time_list)/len(time_list),
        'fastest_time' : min(time_list)
    }
    return summary_dict

acc =[0.95,0.97,0.96]
time=[0.5,0.7,0.6]

print(make_summary(acc,time))

```
