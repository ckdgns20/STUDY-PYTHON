```python
#15번 
k1,p1 = input().split()
k2,p2 = input().split()
k3,p3 = input().split()
k1,k2,k3= int(k1),int(k2),int(k3)
p1,p2,p3= float(p1),float(p2),float(p3)

print(f"총칼로리: {k1+k2+k3}kcal \n총단백질: {p1+p2+p3:.1f}g")
```

```python
#15번 다른 정답
data=[]
del sum
for _ in range(3):
    k,p = input().split()
    data.append((int(k),float(p)))
    
total_kcal = sum(item[0] for item in data)
total_protein = sum(item[1] for item in data)
    
print(f"총칼로리: {total_kcal}kcal \n총단백질: {total_protein:.1f}g")
```
