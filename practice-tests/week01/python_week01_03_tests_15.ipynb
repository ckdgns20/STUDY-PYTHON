```python
#4
code = int(input())
th = code // 1000
remain_num = code % 1000
hun = remain_num // 100
remain_num %= 100
te = remain_num // 10
on = remain_num % 10 
total_sum = th + hun + te + on

print("천의 자리: ", th)
print("백의 자리: ", hun)
print("십의 자리: ", te)
print("일의 자리: ", on)
print("자릿수 합: ", total_sum)
```

```python
#5

coordinate = list(map(int,(input().split())))
x_midpoint = (coordinate[0]+coordinate[2]) / 2
y_midpoint = (coordinate[1]+coordinate[3]) / 2

print(f"중점: ({x_midpoint:.1f},{y_midpoint:.1f})")
```

```python
#6
point = list(map(float,(input().split())))

distance = ((point[2]-point[0])**2 + (point[3]-point[1])**2)**0.5

print(f"두 점 사이의 거리: {distance:.2f}")


```

```python
#7
h_pix, v_pix, n_ch = map(int,(input().split()))

total_pix = h_pix * v_pix
num_total = total_pix * n_ch

print("전체 픽셀 수: ", total_pix)
print("프레임 원소 수: ", num_total)
```

```python
#8
origin_vid = input().split()
h_origin = int(origin_vid[0])
v_origin = int(origin_vid[1])
reduction_ratio = float(origin_vid[2])

after_h = int(h_origin*reduction_ratio)
after_v = int(v_origin*reduction_ratio)
after_resolution = after_h*after_v

print(f"변환 후 해상도: {after_h} × {after_v}")
print("변환 후 픽셀 수:", after_resolution)
```

```python
#9
record = list(map(float,input().split()))
final_record = record[0] * 0.3 + record[1] * 0.5 + record[2] * 0.2

print(f"최종 점수: {final_record:.2f}점")
```

```python
#10
last_acc, new_acc = map(float, input().split())

increase = new_acc - last_acc
delta_acc = (increase / last_acc) * 100

print(f"증가량: {increase:.2f}%p \n변화율: {delta_acc:.2f}%")
```

```python
#11
num_products, per_box = map(int, input().split())

need_boxs = (num_products + per_box -1) // per_box
last_in_box = num_products % per_box

print(f"필요한 박스 수: {need_boxs}개 \n마지막 박스 제품 수: {last_in_box}개")
```

```python
#12
r, g, b = map(int, input().split())

mean_brightness = (r+g+b) / 3

print(f"RGB: ({r}, {g}, {b})")
print(f"평균 밝기: {mean_brightness:.2f}")
```

```python
#13
number_frame, fps = map(int, input().split())

total_sec = number_frame / fps
locate_min = total_sec // 60
locate_sec = total_sec % 60

print(f"시간 위치: {int(locate_min)}분 {locate_sec:.2f}초")
```

```python
#14
A_infer, B_infer = map(float, input().split())

per_A_infer = 1000 / A_infer
per_B_infer = 1000 / B_infer

speed_multi = B_infer / A_infer 

print(f"모델 A 처리량: {per_A_infer:.2f} samples/s \n모델 B 처리량: {per_B_infer:.2f} samples/s \n모델 A는 모델 B보다 {speed_multi:.2f}배 빠릅니다.")
```

```python
#15
log_experiment = input().split()

name_ex = log_experiment[0]
num_train , num_test = map(int, log_experiment[1:3])
acc, infer = map(float, log_experiment[3:5])

total_samples = num_train + num_test
ratio_test = num_test / total_samples * 100

print(f"[{name_ex}] 전체 {total_samples}개 | 테스트 비율 {ratio_test:.1f}% | 정확도 {acc:.1f}% | 추론 {infer:.2f}ms")
```
