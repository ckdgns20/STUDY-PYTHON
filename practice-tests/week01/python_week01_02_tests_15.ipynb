```python
#1번
p,e,d = map(int,input().split())
price = p*e
total_price = price + d
print(f"상품금액: {price}원 \n배송비: {d}원 \n총 결제금액: {total_price}원")
```

```python
#2번
dist, m = map(float, input().split())
h = m/60
velocity = dist/h
print(f"평균 속력: {velocity:.2f}km/h")
```

```python
#3번
r,h = map(int, input().split())
v = 3.14159*r**2*h
print(f"원기둥 부피: {v:.2f}")
```

```python
#4번
p,s,c = map(int, input().split())
after_price = int(p * (s/100))
print(f"할인 적용 금액: {p-after_price}원 \n최종 결제금액: {p-after_price-c}원")
```

```python
#5번
s = int(input())

day = s // (24*60*60)
s %= (24*60*60)
hour = s // 3600
s %= 3600
m = s // 60
s %= 60

print(f"{day}일 {hour}시간 {m}분 {s}초")
    
```

```python
#7번
byte = int(input())
kb = byte // 1024
rest = byte % 1024
print(f"{kb}KB {rest}Byte")
```

```python
#8번
C = float(input())
F = C*1.8 +32
A = C + 273.15
print(f"화씨: {F:.2f}F \n절대온도: {A:.2f}K")
```

```python
#9번 -> 틀림(형 변환은 리스트 전체를 못하고 하나씩 해야함)
#name, kor, eng, mat = input().split()
list1 = list(input().split())
list1[1:] = map(int, list1[1:])
total = sum(list1[1:])
mean = total / len(list1[1:])
print(f"{list1[0]} | 합계: {total}점 | 평균: {mean:.1f}점")
```

```python
#10번
recept = list(input().split())
recept[1:] = map(int, recept[1:])
print(f"{recept[0]} {recept[2]}개 x {recept[1]}원 = {recept[2]*recept[1]}원")
```

```python
#11번
list1 = input().split()
list1[1:] = map(float, list1[1:])
mean = sum(list1[1:])/len(list1[1:])
print(f"{list1[0]} 평균 측정값:{mean:.2f}")
```

```python
#12번 -> 틀림(map은 map 객체가 따로 있어서 list처럼 인덱싱과 슬라이싱 안됌)
budget = list(map(int, input().split()))
total_cost = sum(budget[0:-1])
rest_budget = budget[-1]-total_cost
per = (total_cost / budget[-1]) * 100
print(f"총지출: {total_cost}원 \n남은 예산: {rest_budget}원 \n예산 사용률: {per:.1f}%")
```

```python
#13번
fual = list(map(float,input().split()))
need_fual = fual[0]/fual[1]
percost= need_fual*fual[-1]

print(f"필요 연료: {need_fual:.2f}L \n예상 연료비: {percost:.0f}원")

```

```python
#14번
money = input().split()
principal, years = int(money[0]),float(money[-1])/12
rate = float(money[1])


interest = principal*rate / 100*years
maturity = principal + interest

print(f"이자: {interest:.0f}원 \n만기 금액: {maturity:.0f}원")

```

```python
#15번
data = list(map(int,input().split()))
full_frame = data[0]*data[1]
full_second = full_frame / data[-1]
full_hour = full_second // 3600
full_minute = full_second // 60
full_second %= 60

print(f"전체 프레임: {full_frame} \n예상 처리 시간: {full_hour:.0f}시간 {full_minute:.0f}분 {full_second:.0f}초")
```
