```python
#14
list14 = list(map(int,input().split()))
add_one14 = int(input())

for idx, val in enumerate(list14) :
    if val >= add_one14 :
        list14.insert(idx,add_one14)
        break
else :
    list14.append(add_one14)

print(list14)

```
