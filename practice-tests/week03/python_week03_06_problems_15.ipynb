```python
#5
dict5 = {}
while True:
    try:
        key, value = input().split('=')
        dict5[key] = int(value)
    except ValueError :
        break

print(dict5)
```

```python
#6
acc_class = input().split()

acc_dict6 = dict(zip(acc_class[::2],map(float,(acc_class[1::2]))))

mean_acc = sum(acc_dict6.values())/len(acc_dict6.keys())

print(f"평균 정확도: {mean_acc:.2f}")
```

```python
#9
dict9 = {}
sensor_keys = ['temperature','vibration']
for key in sensor_keys :
    sensor_values = list(map(float,input(f"{key} 측정값 입력하세요: ").split()))
    dict9[key] = sensor_values
    
print(dict9)
    
    
```

```python
#10
dict10 = {}
for sensor, values in dict9.items() :
    dict10[sensor] = sum(values) / len(values)
    
print(dict10)
```

```python
#11
dict11 = {}
while True :
    try:
        datas = input().replace(':',' ').split()
        if datas == []:
            break
    
        product_id = datas[0]
        dict11[product_id] = {'score':float(datas[1]),
                            'result':datas[2]}
    except ValueError :
        break
    
print(dict11)
```

```python
#12
fail_id = []
for id in dict11.keys():
    if dict11[id]['result'] == 'fail' :
        fail_id.append(id)
        
print(fail_id)
```

```python
#13
dict13 = {}
max_acc = 0
max_acc_exp = None
while True:
    result_data = input().replace(':',' ').split()
    if result_data == []:
        break
    
    exp_id = result_data[0]
    dict13[exp_id] = {'accuracy':float(result_data[1]),
                      'inference_ms':float(result_data[2])}
    if dict13[exp_id]['accuracy'] > max_acc :
        max_acc = dict13[exp_id]['accuracy']
        max_acc_exp = exp_id
        
        
    
print(f"최고 정확도 실험: {max_acc_exp}")
```

```python
#14
train_class = set(input().split())
val_class = set(input().split())

if train_class == val_class :
    print("일치")
else:
    if train_class - val_class :
        print(f"train에만 존재: {train_class - val_class}")
    else :
        print(f"validation에만 존재: {val_class-train_class}")
    
```

```python
#15
exp_name = input("실험명 입력")
exp_resol = tuple(map(int,(input("해상도 입력").split())))
class_list = list(input("클래스 목록 입력").split())
configs_inputs = list(map(int,input("설정값 입력").split()))
count_samples = list(map(int,input("샘플 수 입력(normal low high fault 순서대로 입력)").split()))

configs = {'fps':configs_inputs[0],
           'window':configs_inputs[1],
           'stride':configs_inputs[2]}

counts = {'normal':count_samples[0],
          'low':count_samples[1],
          'high':count_samples[2],
          'fault':count_samples[3]}

dict15 = {'name':exp_name,
          'resolution':exp_resol,
          'classes':class_list,
          'unique_classes':set(class_list),
          'config':configs,
          'conts': counts
          }

print(dict15)
```
