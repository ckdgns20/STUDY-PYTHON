```python
#9
setA = set(map(int,input().split()))
setB = set(map(int,input().split()))

total_set9 = setA.union(setB) #setA | setB 도 가능

print(total_set9)
```

```python
#10
setC = set(map(int,input().split()))
setD = set(map(int,input().split()))

inter_set10 = setC.intersection(setD) # setC & setD 도 가능

print(inter_set10)
```

```python
#11
setA11 = set(map(int,input().split()))
setB11 = set(map(int,input().split()))

onlyA = setA11 - setB11

print(onlyA)
```

```python
#12
setA12 = set(map(int,input().split()))
setB12 = set(map(int,input().split()))

sym_dif = setA12 ^ setB12

print(sym_dif)
```

```python
#13
train_data = set(input().split())
test_data = set(input().split())

onlytest = test_data - train_data

print(f"테스트 전용 클래스: {onlytest}")
```

```python
#15
aframe_idx = set(map(int,input().split()))
bframe_idx = set(map(int, input().split()))

both_idx = aframe_idx & bframe_idx
onlyaidx = aframe_idx - bframe_idx
totalidx = aframe_idx | bframe_idx

print(f"공통: {both_idx}")
print(f"A만: {onlyaidx}")
print(f"전체: {totalidx}")
```
