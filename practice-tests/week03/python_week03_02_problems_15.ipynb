```python
#1
list1 = list(map(int,input().split()))

sum1 = 0
for i in list1 :
    sum1 += i

print(f"합계: {sum1}")
```

```python
#2
list2 = list(map(int,input().split()))

sum2 = 0

for i in list2 :
    sum2 += i

avg2 = sum2 / len(list2)

print(f"평균: {avg2:.2f}")
```

```python
#3
list3 = list(map(int,input().split()))
new_list3 = []

for i in list3 :
    if i > 0 :
        new_list3.append(i)
        
print(new_list3)
```

```python
#4
list4 = list(map(int,input().split()))
new_list4 = []

for i, j in enumerate(list4) :
    if j % 2 == 0 :
        new_list4.append(i)
        
print(new_list4)
```

```python
#5
list5 = list(map(int,input().split()))
find_val = int(input())

count_val = 0
for i in list5:
    if i == find_val:
        count_val += 1
        
print(f"등장 횟수: {count_val}")
```

```python
#6
list6 = list(map(int,input().split()))
find_val6 = int(input())

for i, j in enumerate(list6):
    if j == find_val6:
        print(f"위치: {i}")
        break
else :
    print("-1")
        
```

```python
#7
list7 = list(map(int,input().split()))
max_val = 0
for i in list7:
    if i > max_val :
        max_val = i

print(f"최댓값: {max_val}")
    
```

```python
#8
list8 = list(map(int,input().split()))

min_val = float("inf")
min_idx = 0
for i,j in enumerate(list8):
    if j < min_val:
        min_val = j
        min_idx = i

print(f"최솟값: {min_val}")
print(f"위치: {min_idx}")
```

```python
#11
list11 = list(map(int,input().split()))
sort_list11 = []
# sort는 원본을 바꾸고 return값 none, sorted는 새 객체값으로 return
sort_list11 = sorted(list11)

print(f"원본: {list11}")
print(f"정렬본: {sort_list11}")
```

```python
#15
list15 = list(map(float,input().split()))
sum15 = 0
max_val15 = 0
max_idx15 = 0
sorted_list15 = sorted(list15)
for i, j in enumerate(list15):
    sum15 += j
    if j > max_val15 :
        max_val15 = j
        max_idx15 = i

avg15 = sum15/len(list15)

print(f"평균: {avg15:.2f}")
print(f"최고 정확도 위치: {max_idx15}")
print(f"정렬본: {sorted_list15}")
```
