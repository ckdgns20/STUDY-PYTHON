```python
#2
words2 = input().split()
frq2 = {}
### 자료형 확인은 isinstance(변수, 찾는 자료형) or type()

for word in words2 :
    if word in frq2 :
        frq2[word] += 1
    else :
        frq2[word] = 1
    
print(frq2)

```

```python
#3
frq3 = frq2

max_count = 0
max_word = []

for word, count in frq3.items() :
    if count > max_count :
        max_count = count
        max_word = [word]
    elif count == max_count :
        max_word.append(word)
        
print(f"최다 단어: {', '.join(max_word)}")
print(f"횟수: {max_count}")
# 리스트 컴프리헨션 -> [ 변수활용 for 변수 in 반복가능한객체 ]    : 반드시 []안에 넣어줘야함.    
    
```

```python
#4
data4 = input().split()
dict4 = {name: float(score) for name, score in zip(data4[0::2],data4[1::2])}

total_score = 0
for value in dict4.values() :
    total_score += value
    
mean_score =  total_score / len(dict4.keys())

print(f"평균: {mean_score:.2f}")  
```

```python
#5
data5 = input().split()
names = []
scores = []

for i in data5:
    if i.isdigit() :
        scores.append(float(i))
    else:
        names.append(i)

dict5 = dict(zip(names,scores))

for name, score in dict5.items():
    if score >= 90:
        print(name,end=" ")
```

```python
#6
sensor_dict6 = {'temp':[30,40,120], 'pressure':[50,-1,70], 'vibration':[10,20,30]}
count_dict6 = {}

for key, values in sensor_dict6.items() :
    count = 0
    for value in values:
        if 0 < value < 100 :
            count += 1
    
    count_dict6[key] = count
    
print(count_dict6)
```

```python
#7
data_input7 = input().split()

class_dict7 = {}

for key in data_input7 :
    
    if not key in list(class_dict7.keys()) :
        class_dict7[key] = 1
    else :
        class_dict7[key] += 1
    



print(class_dict7)
```

```python
#8
input_data8 = input().split()
keys8 = []
values8 = []


for idx, data in enumerate(input_data8) :
    if idx % 2 == 0 :
        keys8.append(data)
    else :
        values8.append(float(data))

class_dict8 = dict(zip(keys8,values8))

print(class_dict8)
```

```python
#9
dict9 = class_dict8
min_class = None
min_value = float('inf')
for cls, value in dict9.items():
    if value < min_value :
        min_value = value
        min_class = cls
        
print(f"최저 클래스: {min_class}")
print(f"정확도: {min_value:.2f}")
```

```python
#10
input_data10 = input().split()
# Kim 90 AI Lee 85 ME
names = []
scores = []
majors = []

dict10 = {}
for idx, data in enumerate(input_data10):
    if idx % 3 == 0 :
        names.append(data)
    elif idx % 3 == 1 :
        scores.append(data)
    else :
        majors.append(data)
    
for i, name in enumerate(names) :
    dict10[name] = {'score':int(scores[i]),'major':majors[i]}
    
print(dict10)
```

```python
#11
dict11 = dict10

print(dict11['Kim']['score'])
```

```python
#15
input_data15 = input().split()
# A 0.95 0.50 B 0.97 0.65 C 0.94 0.40
dict15 = {}
exp_keys = ['acc','ms']
exp_names = [input_data15.pop(idx) for idx, data in enumerate(input_data15) if not data.isalpha() == 0]
input_data15 = list(map(float,input_data15))
input_data15 = list((zip(input_data15[::2],input_data15[1::2])))

max_acc_exp = None
max_acc = 0
min_ms_exp = None
min_ms = float('inf')
for i, name in enumerate(exp_names) :
    dict15[name] = dict(zip(exp_keys,input_data15[i]))
    acc, ms = input_data15[i]
    if acc > max_acc :
        max_acc = acc
        max_acc_exp = name
    if ms < min_ms :
        min_ms = ms
        min_ms_exp = name
        
print(f"최고 정확도: {max_acc_exp}")
print(f"최고 속도: {min_ms_exp}")
```
